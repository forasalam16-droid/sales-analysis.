```python
df = pd.read_csv(r"C:\Users\HP\Documents\data analysis\FINAL PROJECT\10000 Sales Records.csv")
print(df.head())
```

```python
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["processing_days"]= (df["Ship Date"] - df["Order Date"]).dt.days

```

```python
print(df.processing_days)
```

```python
df["Revenue_per_unit"] = df["Total Revenue"] / df["Units Sold"]
print(df.Revenue_per_unit)
```

```python
df["Cost_per_Unit"] = df["Total Cost"] / df["Units Sold"]
print(df.Cost_per_Unit)
```

```python
df.groupby("Region")[["Total Revenue","Total Profit"]].sum()
```

```python
df.groupby("Item Type")["Total Profit"].sum().sort_values(ascending=False)
```

```python
df.groupby("Sales Channel")["Total Revenue"].sum()
```

```python
df.groupby("Order Priority")["Total Profit"].mean()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
```

```python
df.groupby("Year")["Total Revenue"].sum()
df.groupby("Month")["Total Revenue"].mean()
```

```python
df["Profit_Margin"] = (df["Total Profit"] / df["Total Revenue"]) * 100
print(df.Profit_Margin)
```

```python
df.groupby("Region")["Profit_Margin"].mean()
```

```python
df.groupby("Region")["Total Revenue"].sum().plot(kind="bar")
plt.title("Total Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Total Revenue")
plt.show()
```

```python
df.groupby("Year")["Total Revenue"].sum().plot(kind="line", marker='o')
plt.title("Year-wise Revenue Trend")
plt.xlabel("Year")
plt.ylabel("Total Revenue")
plt.show()
```

```python
df["Sales Channel"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Sales Channel Distribution")
plt.ylabel("")
plt.show()
```

```python
df.boxplot(column="Profit_Margin", by="Region")
plt.title("Profit Margin by Region")
plt.suptitle("")
plt.xlabel("Region")
plt.ylabel("Profit Margin (%)")
plt.show()
```

```python
df.groupby("Item Type")["Total Profit"].sum().sort_values(ascending=False).head(5).plot(kind="bar")
plt.title("Top 5 Item Types by Total Profit")
plt.xlabel("Item Type")
plt.ylabel("Total Profit")
plt.show()
```

```python
df.groupby("Month")["Total Revenue"].mean().plot(kind="bar")
plt.title("Average Revenue by Month")
plt.xlabel("Month")
plt.ylabel("Average Revenue")
plt.show()
```

```python
df["Processing_Days"] = (df["Ship Date"] - df["Order Date"]).dt.days

df["Processing_Days"].plot(kind="hist", bins=20)
plt.title("Order Processing Time Distribution")
plt.xlabel("Processing Days")
plt.ylabel("Frequency")
plt.show()
```

```python
df.groupby("Region")["Total Profit"].sum().plot(kind="bar")
plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
df.groupby("Sales Channel")["Units Sold"].sum().plot(kind="bar")
plt.title("Units Sold by Sales Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Units Sold")
plt.show()
```

```python
plt.scatter(df["Unit Price"], df["Total Profit"])
plt.title("Unit Price vs Total Profit")
plt.xlabel("Unit Price")
plt.ylabel("Total Profit")
plt.show()
```
